# Сбор информации из групп

```python
# кладём токен в переменную для удобства
token = '3f84da623f84da623f84da62b83cc0173733f843f84da62555c4d080b\
40aeda67a4e055'
# для удобства помещаем url в переменную, используем метод протокола для сбора 
# данных
url = 'https://api.vk.com/method/groups.getMembers'
# словарь с параметрами /, нвправляетя в метод get библиотеки requests,
# указываем адрес группы в вконтвке, для того чтобы получить отчет об участниках
# группы
params = {'group_id': 'vk', 'v': '5.95', 'access_token': token}
# получаем данные
response = requests.get(url, params=params)
# десириализируем строку в словарь
data = response.json()
pprint(data)
# узнаём количество элементов в списке с id и 
# утверждаем ограничение на элементов за один запрос
print(len(data['response']['items']))
# создаим переменную, которая хранито двадцать элементов для проверки 
users_for_checking = data['response']['items'][:20]
# создём счетчик count и счетчик offset для цикла, чтобы выгружать данные 
count = 5
offset = 0
# создаётся пустой сисок для храненя элементов
users_id = []
# создаём данные для цикла, чтобы цикл остановился по условию
max_count = 20
# определеям цикл
while offset < max_count:
    print({f'Выгружаю {count} пользователей с offset: {offset}'})
    params = {'group_id': 'vk', 'v': '5.95','count': count, 'offset': offset, \
    'access_token': token}
    response = requests.get(url, params=params)
    data = response.json()
    users_id += data['response']['items']
    offset += count
# сравним списки полученные двумя способами
print(users_for_checking == users_id)
```

# Лайки репосты и комментарии


```python
# получаем данные со стены вконтаке, используем необходимый метод и 
# фильтруем по владельцу сообщений
token = '3f84da623f84da623f84da62b83cc0173733f843f84da62555c4d080b\
40aeda67a4e055'
url = 'https://api.vk.com/method/wall.get'
params = {'domain': 'vk', 'filter': 'owner', 'count': 1000, 'offset': 0, \
'access_token': token, 'v': 5.95 }
response = requests.get(url, params=params)
data = response.json()

print(len(data['response']['items']))
#pprint(data['response']['items'][0])
pprint(data)
```

```python
# создём пустой словарь с для данных
stats = {}
# создаём счетчик ограничение
max_count = 0
# проходимся по данным циклом 
for record in data['response']['items'][:]:
    # проверям что сообщение не пустое
    if record['text'][:30]:
        stats[record['text'][:30]] = [record['comments']['count'], \
        record['likes']['count'], record['reposts']['count'], record['date']]
        # увеличиваем счетчик на 1
        max_count += 1
    if max_count < 10:
        continue
    else:
        break
pprint(stats)
```
