```python
# обязательный импотр библиотеки
import requests
import json
from pprint import pprint
```

```python
# в переменную направляется url адрес
url = 'https://www.cbr-xml-daily.ru/daily_json.js'
# в переменную направляется инофрмация о курсе валюты цбрф 
# и используется метод .get()
response = requests.get(url)
# получаем обьект Responce и успешный статус запроса
print(response)
# получаем информацию только о статсуе запроса
print(response.status_code)
```

```python
# смотрим на содердимое переменной при помощи атрибута text
print(response.text)
# десириализуем строку в словарь
# используется втроенный в библотеку request метод .json()
currencies = response.json()
pprint(currencies)
# выведем на экран информацию о евро
pprint(currencies['Valute']['EUR'])
```
