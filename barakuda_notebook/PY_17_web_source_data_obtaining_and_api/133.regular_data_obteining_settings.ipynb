# Постановака задачи и выполнение функции

`.every()`

```python
# предварительно очищаем список задач к выполнению
schedule.clear()
def task():
    print('Hello i am task!')
    return
# создаётся автоматическое включение скрипта
#schedule.every(1).hour.do(task)
schedule.every(1).minutes.do(task)
# создётся скрипт в бесконечном цикле, 
# который проверяет наличие задач к выполнению
while True:
    schedule.run_pending()
    # не перегружаем компьютер, действия повторяются каждую секунду
    time.sleep(1)
```
