# Задание 6.2 

Используя API, определите долю женщин (sex=1) среди пользователей с id от 1 до 500. Иногда будут попадаться пользователи, у которых пол не указан (sex=0), — таких пользователей не нужно учитывать в общем числе.

В ответе укажите число, округлив до двух знаков после точки-разделителя, например, 0.55.

```python
# создаем список из цифр, для того чтобы использовать его в значении параметра
ids = ','.join(map(str, range(1, 500)))
# записываем в переменную идентификатор, токен вконтаке
token = '3f84da623f84da623f84da62b83cc0173733f843f84da62555c4d080b40\
aeda67a4e055'
# записываем в переменную наименование веб сайта
url = 'https://api.vk.com/method/users.get'
# используем параметры для анализа требуемой информации
params = {'user_ids': ids, 'v': 5.95, 'fields': 'sex,bdate', \
'access_token': token, 'lang': 'ru'}
# сразу печатаем, получаем и десирализируем данные
#pprint(requests.get(url, params=params).json())
users = requests.get(url, params=params).json()['response']
# создаём счетчик для получения значения количеста женского и мужского пола
female = 0
male = 0
# создаём цикл для рассмотрения параметра пола и обработки дынных
for elem in users:
    # условиями прохожу по отчету о клиентах
    if elem['sex'] == 1:
        female += 1
    elif elem['sex'] == 2:
        male +=1     
# печатаем ответ
print(f'Доля женщин среди пользователей с id jn 1 до 500: {round(female / \
(female + male), 2)}')
pprint(users)
```
