Устанавливаем библиотеку BeautifulSoup `sudo apt install python3-bs4`

```python
# импорт необходимой библиотеки
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://nplus1.ru/news/2021/10/11/econobel2021'
# обычным образом получаем данные html расширения в переменную 
response = requests.get(url)
# соззаём обьект BeautifulSoup с парамсетром html.parser для того чтобы 
# орабатывать информацию веб страницы
page = BeautifulSoup(response.text, 'html.parser')
# получаем выбранный тег title
print(page.title)
# выводим текст по атрибуту text
print(page.title.text)
```

# Извлекаем заголовок статьи

`.find()`

```python
# применяем к обьеку BeautifulSoup метод для того чтобы получить текс тега
print(page.find('h1').text)
```

# Неуникальные теги: извлекаем текст и дату публикации статьи

```python
# смотрим на данные статьи с страницы и используем название класса в параметрах 
# для корректного отбражения требуемого тега
print(page.find('div', class_='n1_material text-18').text)
print(page.find('a', class_='relative before:block before:w-px before:bg-\
current before:h-4 before:absolute before:left\
-0 group pl-2 flex inline-flex items-center').text)
# знаем что можно корректно сокращать название тега при поиске
print(page.find('a', class_='relative').text)
```

# Сбор нескольких элементов: собираем все ссылки на старнице

```python
# Задаём адрес
url = 'https://en.wikipedia.org/wiki/List_of_programming_languages'
# получаем данные, используется метод при котором исключается статаус 403
response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
# создаём объект BeautifulSoup для анализа веб страницы, html расширения
page = BeautifulSoup(response.text, 'html.parser')
# ищем данные по тегу, который предназначен для содержания ссылок
print(page.find('a'))
# испльзуем метод для получения всех ссылок на странице
links = page.find_all('a')
# смотрим на количество наименований ссылок в списке
print(len(links))
# смотрим на часть ссылок
print([link.text for link in links[500: 510]])
print([link.text for link in links[0: 10]])
```
