Обьеденяем таблицу по ключу
```
SELECT
FROM
    sql.teams
    sql.matches
WHERE home_team_api_ip = api_id
```

```
SELECT
    long_name,
    home_team_goals,
    away_team_goals
FROM
    sql.teams
    sql.matches
WHERE home_team_api_id = api_id
```
