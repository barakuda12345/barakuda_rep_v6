`INNER JOIN`

Отображаем соответстиве значениям в таблицах и увидим, что некоторвые записи одной таблицы отсутствуют в другой
```
SELECT
    COUNT(DISTINCT t.id)
FROM sql.teams t
JOIN sql.matches m ON t.api_id = m.home_team_api_ai OR t.api_id = away_team_api_id
```


`LEFT OUTER JOIN` и `RIGHT OUTER JOIN`

В варианте соединения таблиц методом LEFT, даннве из  правой таблицы, которые не соответствую условию
принимают значение NULL
```
SELECT
    t.long_name
    m.id
FROM sql.teams t
LEFT JOIN sql.mathes m ON t.api_id = m.away_team_api_id OR t.api_id = m.away_team_api_id
ORDER BY m.id DESC
```

Смотрим на названиен команд с пустыми запсиями
```
SELECT
    t.long_name
FROM sql.teams t
LEFT JOIN sql.mathes m ON t.api_id = m.away_team_api_id OR t.api_id = m.away_team_api_id
WHERE m.id is NULL
```

Уиверждаем, что для пустая запись не отображаются и вызывается ошибка
```
SELECT
    t.long_name
FROM sql.teams t
LEFT JOIN sql.mathes m ON t.api_id = m.away_team_api_id OR t.api_id = m.away_team_api_id
WHERE 
    m.season = '2008/2009'
AND t.long_name = 'KAS Eupen'
```

Используем агрегатную функцию и утвержаем, что все агрегатные функции кроме count считаются как NULL
```
SELECT
    t.long_name
    SUM(away_team_api_id) total goals
FROM sql.teams t
LEFT JOIN sql.mathes m ON t.api_id = m.away_team_api_id
GROUP BY t.id
ORDER BY 2 DESC
```

`FULL OUTER JOIN`

Утвержаем что при соединении добвалются несоответствующие записи з обеих таблиц
```
SELECT
...
FROM table_one
FULL OUTER JOIN table_two ON условие
```

`CROSS JOIN`

Отображаем декартовое произведение
```
SELECT *
FROM
    sql.teams
    CROSS JOIN sql.matches
```

Отображаем второй вариант декартового произведения
```
SELECT *
FROM
    sql.teams
    INNER JOIN sql.matches ON True
```

Получаем данные о всех возможниых комбинациях команд
```
SELECT
    DISTINCT
    t_one.long_name home_team
    t_two.long_name away team
FROM
    sql.team t_one
    CROSS JOIN sql.team t_two
```

`NATURAL JOIN`


Отображается синтаксис при котором соединение происходит по всем столбцам столбцам с одинаковыми именами
```
SELECT *
FROM table_one
NATURAL JOIN table_two
```
