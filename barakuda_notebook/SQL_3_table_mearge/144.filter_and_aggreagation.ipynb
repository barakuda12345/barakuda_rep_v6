Выбираем данные для отображения
```
SELECT
    m.id
FROM
    sql.teams t
    JOIN sql.matches m ON m.away_team_api_id = t.api_id
WHERE long_name = 'Arsenal'
```


Отобажаем столбцы из двух таблиц
```
SELECT
    m.id id_1
    m.season,
    t.id id_2
    t.long_name
FROM 
    sql.teams t
    JOIN sql.matches m ON m.away_team_api_id = t.api_id
WHERE long_name = 'Arsenal'
```

Фильтруем по двум условиям
```
SELECT *
FROM 
    sql.matches m
    JOIN sql.teams t ON t.api_id = m.home_team_api_id
WHERE
    t.short_name = 'GEN'
    AND m.season = '2008/2009'
```

Применяется агрегирующая функция
```
SELECT
    t.long_name
    SUM(m.home_team_goals) + SUM(m.away_team_goals) match_goals
FROM
    sql.matches m
    JOIN sql.teams t ON m.away_team_api_id = t.api_id
GROUP BY t.id
```

Отображаем данные о команде в сезоне с сумарным количеством голов больше 100
```
SELECT
    m.season
    t.long_name
    SUM(m.home_team_goals) + SUM(m.away_team_goals) totoal_goals
FROM
    sql.matches m
JOIN sql.teams t ON m.home_team_api_id = t.api_id OR m.away_team_api_id = t.api_id
GROUP BY m.season,t.id
HAVING SUM(m.home_team_goals) + SUM(m.away_team_goals) > 100
```
