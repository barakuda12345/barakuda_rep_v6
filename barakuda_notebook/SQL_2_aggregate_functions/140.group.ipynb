`GROUP BY`

Отображаем результат группировкаи и агрегирующей функции
```
SELECT
    type1 AS pokemon_type,
    COUNT (*) AS pokemon_count
FROM sql.pokemon
GROUP BY type1
ORDER BY type2
```

Смотрим на пример сортировки
```
SELECT
    type1 AS pokemon_type,
    COUNT (*) AS pokemon_count
FROM sql.pokemon
GROUP BY type1
ORDER BY COUNT(*) DESC
```

Групируем с использование сразу двух столбцов и сортируем, так чтобы в начале таблицы стояли значения с нулями
```
SELECT
    type1 AS pokemon_type,
    type2 AS additional_type
    COUNT (*) AS pokemon_count
FROM sql.pokemon
GROUP BY 1, 2
ORDER BY 1, 2 NULL FIRST
```

Отобажаем сгруппированнцю таблицу без агрегирующих функций
```
SELECT
    type1 
FROM sql.pokemon
GROUP BY type1
```
