`COUNT`

Считаем количество строк во всех столбцах
```
SELECT
    COUNT(*)
FROM sql.pokemon
```

Вычисляем количеситво строк столбца с ункальными знчениями
```
SELECT 
    COUNT(DISTINCT type1)
FROM sql.pokemon
```

Отображаем данные при участии нескольких агрегирующих значений
```
SELECT 
    COUNT(*) AS 'Всего трявяных покемонов'
    COUNT(type2) AS 'Покемонов с дополнительным типом'
    AVG(attack) AS 'Средняя атака'
    AVG(defense) AS 'Средняя защита'
FROM sql.pokemon
WHERE type1 = 'Grass'
```
