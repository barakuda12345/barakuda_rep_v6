`UNION ALL`

Корректно обьединяем таблицы, так чтобы количество, очередность и тип столбцов совпадал.
```
SELECT book_name object_name, 'книга' object_discription
FROM public.books
UNION ALL
SELECT movie_title, 'фильм'
FROM sql.kinopoisk
```

Объеденяем таблицы и учитываем, что название столбцов будет соответствовать двнным первой выбранной таблицы
```
SELECT
    c.city_id object_name, 'id города' object_type
FROM
    sql.city c
UNION ALL
SELECT
    c.driver_id other_name, 'id водителя' other_type
FROM
UNION ALL
    sql.driver d
SELECT
    c.ship_id, 'id доставки' 
FROM
    sql.shipment s
UNION ALL
SELECT
    c.cust_id, 'id клиента' 
FROM
    sql.customer c
UNION ALL
SELECT
    c.truck_id, 'id грузовик' 
FROM
    sql.truck s
UNION ALL
ORDER BY 1
```

Сортиреуется и ограничивается каждая часть объединенной талицы 
```
(SELECT 
    book_name object_name, 'книга' object_discription
FROM
    public.books
ORDER BY 1
LIMIT 1)
UNION ALL 
(SELECT 
    movie_title, 'фильм'
FROM
    sql.kinopoisk
ORDER BY 1
LIMIT 1)
```
