```python
# импортируем модуль из пакета
import xml.etree.ElementTree as ET
import pandas as pd
```

# Извлекаем объект из xml-файла

```python
# считаем содержимое файла в переменную
tree = ET.parse('menu.xml')
```

# Корень

```python
# Запишем в переменную root корневой узел
root = tree.getroot()
display(root)
# знаем тип обьекта
display(type(root))
```

# Потомки

```python
# получаем данные о потомках в корневом узле
display(list(root))
# получаем данные списке блюд второго потомка
display(list(root[1]))
```

# Атрибуты и теги

```python
# получаем данные об атрибуте
display(root[0].attrib)
# получаем данные об узле
display(root[0][0])
# получаем значение узла
display(root[0][0].text)
# получаем данные о наименовании тега
display(root.tag)
print(f'Наименование узла root[0][2]: {root[0][2].tag}')
```

# Использование циклов

```python
# с помощью цикла получаем данные о лисьях древовидной структуры
for dish in root:
    for param in dish:
        print(dish.attrib['name'], param.tag, param.text)
```

# Загружаем данные из xml-файла в dataframe

```python
# создаём список с именам колон для датафрейм объекта
columns_names = ['name', 'price', 'weight', 'class']
# в спиосок будем перемещать данные для датафрейм объекта
df_list = []
# обходим циклом и получаем данные
for dish in root:
    row = [dish.attrib['name'], dish[0].text, dish[1].text, dish[2].text]
    # готовые списки пермещаем в датафрейм
    df_list.append(row)
# перемещаем в переменную датафрейм структуру
df_dish = pd.DataFrame(df_list, columns=columns_names)
display(df_dish)
```

# Создаём xml-файл

```python
# воссоздаём структуру xml-файла и создаём корень дерева
new_root = ET.Element('menu')
display(new_root)
# добавляем потомков
dish_one = ET.SubElement(new_root, 'dish', name='Кура')
dish_two = ET.SubElement(new_root, 'dish', name='Греча')
display(list(new_root))
# добавляем к узлам потомков, добавляем значение атрибутов
price_one = ET.SubElement(dish_one, 'price').text = '40'
weight_one = ET.SubElement(dish_one, 'weight').text = '300'
class_one = ET.SubElement(dish_one, 'class').text = 'Мясо'
price_two = ET.SubElement(dish_two, 'price').text = '20'
weight_two = ET.SubElement(dish_two, 'weight').text = '200'
class_two = ET.SubElement(dish_two, 'class').text = 'Крупа'
# с помощью цикла получаем данные о лисьях древовидной структуры, 
# утверждаем воссоздание структуры
for dish in root:
    for param in dish:
        print(dish.attrib['name'], param.tag, param.text)
```

# Сохранение xml-файла

```python
# зпишем объект типа ElementTree.Element в строку с помощью метода .tostring()
new_root_string = ET.tostring(new_root)
with open('new_menu.xml', 'wb') as f:
    f.write(new_root_string)
# записываем подготовленный обьект в xml расширение, другим методом
ET.ElementTree(new_root).write('new_menu_good.xml', encoding='utf-8')
```
