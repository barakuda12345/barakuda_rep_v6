# Считываем данные из файла Excel

````python
# загружаем excel-файл в обьект DataFrame и выводим на экран
grades = pd.read_excel('data/grades.xlsx')
display(grades)
````

# Считываение данных из файла по excel сыллке

````python
# загружаем данные по ссылке
data = pd.read_excel('https://github.com/asaydn/test/raw/master/january.xlsx',\
skiprows=3)
# выводим DataFrame на экран
display(data)
````

````python
# выбираем лист из excel книги
data = pd.read_excel('data/grades.xlsx', sheet_name='ML')
# выводим обьект DataFrame на экран
display(data)
# читаем первый другой лист из той же книги
data = pd.read_excel('data/grades.xlsx', sheet_name='Maths')
display(data)
````

# Выгрузка данных из DataFrame в Excel-файл

````python
# создадим Excel файл
grades.to_excel('data/grades_new.xlsx')
````

````python
# передаем в метод название для единственного листа и не сохраняем индексы 
grades.to_excel('data/grades_new.xlsx', sheet_name='Example', index=False)
````
