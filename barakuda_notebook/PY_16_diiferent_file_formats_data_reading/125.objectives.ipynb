```python
# импортируем необъодимые библиотеки
import pandas as pd
```

Считайте данные из двух листов файла ratings+movies.xlsx в разные DataFrame, объедините в один, запишите данные из полученного DataFrame в файл. Сколько строк (включая строку заголовков) в результирующем файле?

```python
# загружаем в переменные разные листы одной книги и соединяем их и при этом мы 
# не учитваем заголовки
ratings = pd.read_excel('data/ratings_movies.xlsx', sheet_name='ratings')
movies = pd.read_excel('data/ratings_movies.xlsx', sheet_name='movies')
# уточняем, что таблицы прочитались корректно
display(ratings)
display(movies)
ratings_movies = ratings.merge(
    movies,
    on='movieId',
    how='left'
)
# отображаем обьединенную таблицу
display(ratings_movies)
# выграужем данные в Excel файл
ratings_movies.to_excel('data/ratings_movies_new.xlsx',)
# читаем данные в пересенную, не учитывая заголовки
ratings_movies_df = pd.read_excel('data/ratings_movies_new.xlsx', header=None)
# выводим на экран информацию о количестве строк
print(f'Количество строк в таблице: {ratings_movies_df.shape[0]}')
```
