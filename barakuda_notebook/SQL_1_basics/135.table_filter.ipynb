`WHERE`

Получаем инофмацию о перво записи в таблице
```
SELECT
FROM sql.kinopoisk
WHERE position = 1
```

Утверждаем использование условных операторов
```
SELECT
    position,
    movie_title,
    year,
    director
FROM sql.kinopoisk
WHERE year < 1984
```

Выводим все данные учитвая исключение
```
SELECT *
FROM sql.kinopoisk
WHERE year <> 2000
```

`AND` и `OR`

Добавляем условие.
```
SELECT *
FROM sql.kinopoisk
WHERE year >= 2000
AND raiting >= 8
```

Получаем информацию о интервале
```
SELECT *
FROM sql.kinopoisk
WHERE year >= 1975
    AND year <= 1985
```

`BETEWEEN`

Оптимизированная запись для получения интервала
```
SELECT *
FROM sql.kinopoisk
WHERE year BETWEEN 1975 AND 1985
```

`NOT`

Получаем информацию о фильмах, которые не входят в указнный промежуток
```
SELECT *
FROM sql.kinopoisk
WHERE year NOT BETWEEN 1975 AND 1985
```

Для соответствия приоритетности используются скобки
```
SELECT *
FROM sql.kinopoisk
WHERE (raiting > 8.5 AND year < 2000)
    OR year >= 2000
```

`IN`

Утверждаем упрощенную коутркцию без OR
`column IN (value_one, value_two, value_three)`

Смотрим на фильтрацию в варианте, когда используется текст
```
SELECT *
FROM sql.kinopoisk
WHERE director = 'Леонид Гайдай'
```

`LIKE`

Находим данные в значения которых начинаются с А
```
SELECT *
FROM sql.kinopoisk
WHERE movie_title LIKE A%
```

`NULL`

Получаем таблицу в которой выбранный столбец только из пустых значений
```
SELECT *
FROM sql.kinopoisk
WHERE overview IS NULL
```
