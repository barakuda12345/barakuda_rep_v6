`ORDER BY`

Выводим отсортированную таблицу
```
SELECT *
FROM sql.kinopoisk
ORDER BY movie_title
```

Смотрим на второй вариант сортировки
```
SELECT *
FROM sql.kinopoisk
ORDER BY raiting ASC
```

Фильтруем и сортируем таблицу 
```
SELECT *
    movie_title,
    director,
    screenwriter,
    year
FROM sql.kinopoisk
WHERE country = 'СССР'
ORDER BY raiting ASC
```

При сортировке получаем пустые значения в начале
```
SELECT *
    movie_title,
    director,
    screenwriter,
    year
FROM sql.kinopoisk
ORDER BY overview NULLS FIRST
```

Задаём несколько параметров сортировки
```
SELECT *
    movie_title,
    director,
    screenwriter,
    year
FROM sql.kinopoisk
ORDER BY year, raiting DESC
```

Для упрощения используем номера из вывода
```
SELECT *
    director,
    movie_title,
    year
FROM sql.kinopoisk
ORDER BY 1, 3 DESC
```
