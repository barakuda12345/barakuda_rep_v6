# Простые операции с данными

Используется для отображения таблицы
```
SELECT *
FROM sql.kinopoisk
```

В запросе создаётся два столбца и третий в качестве результата вычитания. 
Виртуальному стобцу начзначается имя.
```
SELECT
    director,
    movie_title ,
    10 - raiting AS difference
FROM sql.kinopoisk
```

Утверждаем, что можно делить столбцы друг на друга
```
SELECT
    movie_title,
    year / raiting
FROM sql.kinopoisk
```
