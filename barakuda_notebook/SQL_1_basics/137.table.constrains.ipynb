`LIMIT`

Отображение первых 10 строк таблицы
```
SELECT
FROM sql.kinopoisk
LIMIT 10
```

Выводим первые пять строк предварительно отсортировав таблицу
```
SELECT
    movie_title,
    rating
FROM sql.kinopoisk
ORDER BY rating DESC
LIMIT 10
```

`OFFSET`

Смотрим на работу ключевого слова, которое обрезает первые 3 строчки
```
SELECT
    movie_title,
    rating
FROM sql.kinopoisk
ORDER BY rating DESC
OFFSET 3 LIMIT 5
```
